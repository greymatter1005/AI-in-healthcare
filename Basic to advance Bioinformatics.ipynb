Load a DNA Sequence (FASTA format)

What is a k-mer?

A k-mer is a substring of length k taken from a biological sequence (DNA, RNA, protein).

Example:
Sequence = "ATCGAT"

2-mers (k=2): AT, TC, CG, GA, AT

3-mers (k=3): ATC, TCG, CGA, GAT


Example Walkthrough

Sequence = "ATCGAT", k=2

Step 1: Extract kmers → ["AT","TC","CG","GA","AT"]

Step 2: Count → { "AT":2, "TC":1, "CG":1, "GA":1 }

Step 3: Sort keys → ["AT","CG","GA","TC"]

Step 4: Feature vector → [2,1,1,1]

So the sequence "ATCGAT" is now represented as the numeric feature vector [2,1,1,1].

```python
# Robust k-mer -> ML pipeline (toy dataset)
from itertools import product
from collections import Counter
import numpy as np
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

# ---------- Helper functions ----------
def generate_possible_kmers(k):
    """Return sorted list of all k-mers from alphabet ATCG (consistent ordering)."""
    return [''.join(p) for p in product('ATCG', repeat=k)]

def sequence_to_kmer_counts(seq, k, possible_kmers):
    """Return fixed-length vector of counts for all possible_kmers.
       Non-ATCG kmers are ignored (counted as 0)."""
    seq = seq.upper()
    n = len(seq) - k + 1
    if n <= 0:
        return np.zeros(len(possible_kmers), dtype=float)
    kmers = [seq[i:i+k] for i in range(n)]
    # filter invalid kmers containing other letters
    kmers = [km for km in kmers if all(ch in 'ATCG' for ch in km)]
    counts = Counter(kmers)
    return np.array([counts.get(km, 0) for km in possible_kmers], dtype=float)

def sequence_to_kmer_freq(seq, k, possible_kmers):
    """Return normalized frequency vector (sum to 1 if there are valid kmers)."""
    counts = sequence_to_kmer_counts(seq, k, possible_kmers)
    s = counts.sum()
    return counts / s if s > 0 else counts

def build_feature_matrix(sequences, k):
    possible_kmers = generate_possible_kmers(k)
    X = np.vstack([sequence_to_kmer_freq(s, k, possible_kmers) for s in sequences])
    return X, possible_kmers

# ---------- Example dataset ----------
sequences = [
    "ATCGATCGA", "GGGGGGGGG", "ATATATATAT", "CCCCCCCC",
    "ATCGTATCGA", "GGGCGGGGG", "ATATCGATAT", "CCCGCCCC"
]
labels = [1, 0, 1, 0, 1, 0, 1, 0]  # 1 = Disease, 0 = Healthy

# ---------- Build features ----------
k = 2
X, possible_kmers = build_feature_matrix(sequences, k)
y = np.array(labels)

print("Feature matrix shape:", X.shape)         # (n_samples, n_features)
print("Number of features (k-mers):", len(possible_kmers))
print("k-mer order (first 10):", possible_kmers[:10])

# ---------- Train / evaluate ----------
# Use stratify to keep class ratio in train/test
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42, stratify=y
)

# Optional: scaler (k-mer frequencies already 0..1 but scaling can help some models)
scaler = StandardScaler()
X_train_s = scaler.fit_transform(X_train)
X_test_s  = scaler.transform(X_test)

model = LogisticRegression(max_iter=1000, solver='liblinear')  # good for small data
model.fit(X_train_s, y_train)

y_pred = model.predict(X_test_s)
print("\nAccuracy:", accuracy_score(y_test, y_pred))
print("\nConfusion Matrix:\n", confusion_matrix(y_test, y_pred))
print("\nClassification Report:\n", classification_report(y_test, y_pred))

# Cross-validation for more reliable estimate (since dataset is tiny)
cv_scores = cross_val_score(model, scaler.transform(X), y, cv=4, scoring='accuracy')
print("Cross-val accuracy (cv=4):", cv_scores, "mean:", cv_scores.mean())

# ---------- Predict new sample ----------
new_seq = "ATCGTATCGA"
new_X = sequence_to_kmer_freq(new_seq, k, possible_kmers).reshape(1, -1)
new_X_s = scaler.transform(new_X)
pred_label = model.predict(new_X_s)[0]
pred_proba = model.predict_proba(new_X_s)[0, 1]  # probability of class 1

print(f"\nNew sequence prediction: {'Disease' if pred_label==1 else 'Healthy'} "
      f"(probability of disease = {pred_proba:.3f})")
```

CNN for DNA Classification (with step outputs)

```python
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv1D, MaxPooling1D, Flatten, Dense, Embedding
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from tensorflow.keras.utils import to_categorical

# ----------------------------
# Step 1: Toy DNA dataset
# ----------------------------
sequences = [
    "ATCGATCG", "GGGGGGGG", "ATATATAT", "CCCCCCCC",
    "ATCGTACG", "GGGCGGGG", "TATATATA", "CCCGCCCC"
]
labels = [1, 0, 1, 0, 1, 0, 1, 0]   # 1 = Disease, 0 = Healthy

print("Sequences:", sequences)
print("Labels:", labels)

# ----------------------------
# Step 2: Encode DNA as numbers
# ----------------------------
mapping = {"A":0, "T":1, "C":2, "G":3}

encoded_sequences = []
for seq in sequences:
    encoded = [mapping[base] for base in seq]
    encoded_sequences.append(encoded)

encoded_sequences = np.array(encoded_sequences)
labels = np.array(labels)

print("\nEncoded Sequences (Numeric):\n", encoded_sequences)

# ----------------------------
# Step 3: Train-test split
# ----------------------------
X_train, X_test, y_train, y_test = train_test_split(
    encoded_sequences, labels, test_size=0.25, random_state=42
)

print("\nTraining Data Shape:", X_train.shape)
print("Testing Data Shape:", X_test.shape)

# ----------------------------
# Step 4: One-hot encode sequences
# ----------------------------
# Each base → [1,0,0,0] for A, etc.
X_train_oh = to_categorical(X_train, num_classes=4)
X_test_oh = to_categorical(X_test, num_classes=4)

print("\nOne-hot Encoded Shape (Train):", X_train_oh.shape)
print("Example One-hot Encoded Sequence:\n", X_train_oh[0])

# ----------------------------
# Step 5: Build CNN
# ----------------------------
model = Sequential()
model.add(Conv1D(filters=16, kernel_size=3, activation='relu', input_shape=(X_train_oh.shape[1], 4)))
model.add(MaxPooling1D(pool_size=2))
model.add(Flatten())
model.add(Dense(8, activation='relu'))
model.add(Dense(1, activation='sigmoid'))  # Binary classification

model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

print("\nModel Summary:")
model.summary()

# ----------------------------
# Step 6: Train model
# ----------------------------
history = model.fit(X_train_oh, y_train, epochs=10, batch_size=2, verbose=1, validation_data=(X_test_oh, y_test))

# ----------------------------
# Step 7: Evaluate
# ----------------------------
loss, acc = model.evaluate(X_test_oh, y_test)
print("\nFinal Test Accuracy:", acc)

# ----------------------------
# Step 8: Prediction for new DNA sequence
# ----------------------------
new_seq = "GGGGGGGG"
new_encoded = np.array([[mapping[base] for base in new_seq]])
new_encoded_oh = to_categorical(new_encoded, num_classes=4)

prediction = model.predict(new_encoded_oh)
print("\nPrediction for new sequence:", "Disease" if prediction[0][0] > 0.5 else "Healthy")
```

Full code

```python
from Bio import Entrez, SeqIO
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

# ---- Step 1: Fetch 3 sequences from NCBI ----
Entrez.email = "your_email@example.com"

ids = {
    "Human_alpha_HBA1": "NP_000508.1",
    "Mouse_alpha_Hba": "NP_032244.2",
    "Human_beta_HBB": "NP_000509.1"
}

records = {}
for label, uid in ids.items():
    handle = Entrez.efetch(db="protein", id=uid, rettype="fasta", retmode="text")
    record = SeqIO.read(handle, "fasta")
    handle.close()
    records[label] = record

# ---- Step 2: Feature extraction (AA composition) ----
def aa_comp(seq):
    aa = "ACDEFGHIKLMNPQRSTVWY"
    seq = str(seq).upper()
    length = len(seq)
    return [seq.count(a)/length for a in aa]

X = []
y = []
for label, rec in records.items():
    X.append(aa_comp(rec.seq))
    if "Human_alpha" in label:
        y.append("Human_alpha")
    elif "Human_beta" in label:
        y.append("Human_beta")
    else:
        y.append("Mouse_alpha")

df = pd.DataFrame(X, columns=list("ACDEFGHIKLMNPQRSTVWY"))
df['label'] = y

print("\nDataset:\n", df)

# ---- Step 3: Train & test (self-test since data is tiny) ----
X = df.drop(columns='label')
y = df['label']

models = {
    "SVM (linear)": SVC(kernel='linear'),
    "Random Forest": RandomForestClassifier(n_estimators=100, random_state=42),
    "Logistic Regression": LogisticRegression(max_iter=1000, random_state=42)
}

accuracies = {}

for name, model in models.items():
    model.fit(X, y)
    preds = model.predict(X)
    acc = accuracy_score(y, preds)
    accuracies[name] = acc
    print(f"{name} Accuracy: {acc:.2f}")

# ---- Step 4: Plot accuracy comparison ----
plt.figure(figsize=(6,4))
plt.bar(accuracies.keys(), accuracies.values(), color=['skyblue','lightgreen','salmon'])
plt.ylabel("Accuracy")
plt.title("Model Accuracy Comparison on Hemoglobin Sequences")
plt.ylim(0, 1.1)
for i, acc in enumerate(accuracies.values()):
    plt.text(i, acc + 0.02, f"{acc:.2f}", ha='center')
plt.show()
```
