Step 1: Import Libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

Step 2: Create Dataset

```python
# Input (X) and Output (Y)

X = np.array([1,2])

Y = np.array([3,5])

df = pd.DataFrame({
    "Input (X)":X,
    "Output (Y)":Y
})

print(df)
```

Step 3: Initialize Parameters

```python
# Initial values

w = 0
b = 0

learning_rate = 0.1

n = len(X)

print("Initial Weight =",w)
print("Initial Bias =",b)
```

Step 4: Gradient Descent

```python
iterations = 4

for i in range(iterations):

    print("="*60)
    print("Iteration",i+1)

    # Prediction

    y_pred = w*X+b

    print("\nPredictions")

    for j in range(n):

        print(f"x={X[j]} --> y_hat={y_pred[j]:.3f}")

    # Errors

    error = Y-y_pred

    print("\nErrors")

    for j in range(n):

        print(f"e{j+1} = {Y[j]} - {y_pred[j]:.3f} = {error[j]:.3f}")

    # Cost

    mse = np.mean(error**2)

    print("\nMean Squared Error =",round(mse,4))

    # Gradients

    dw = (-2/n)*np.sum(X*error)

    db = (-2/n)*np.sum(error)

    print("\nGradient Calculation")

    print("dw =",round(dw,4))

    print("db =",round(db,4))

    # Update

    w = w-learning_rate*dw

    b = b-learning_rate*db

    print("\nUpdated Parameters")

    print("Weight =",round(w,4))

    print("Bias =",round(b,4))
```

Step 5: Final Regression Line

```python
plt.figure(figsize=(6,4))

plt.scatter(X,Y,color='red',s=80,label="Actual Data")

x_line = np.linspace(1,2,100)

y_line = w*x_line+b

plt.plot(x_line,y_line,label="Regression Line")

plt.xlabel("Input (X)")
plt.ylabel("Output (Y)")
plt.title("Linear Regression using Gradient Descent")
plt.legend()

plt.grid(True)

plt.show()
```

Step 6: Predict New Value

```python
new_x = 10

prediction = w*new_x+b

print("Input :",new_x)

print("Predicted Output :",round(prediction,3))
```

Step 7: Display the Loss Reduction

```python
w = 0
b = 0

losses = []

iterations = 20

for i in range(iterations):

    y_pred = w*X+b

    error = Y-y_pred

    mse = np.mean(error**2)

    losses.append(mse)

    dw = (-2/n)*np.sum(X*error)

    db = (-2/n)*np.sum(error)

    w = w-learning_rate*dw

    b = b-learning_rate*db

plt.figure(figsize=(7,4))

plt.plot(range(1,iterations+1),losses,marker='o')

plt.xlabel("Iteration")

plt.ylabel("Mean Squared Error")

plt.title("Loss Reduction using Gradient Descent")

plt.grid(True)

plt.show()
```
